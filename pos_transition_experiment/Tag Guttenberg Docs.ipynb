```python
import pandas as pd
import numpy as np
import spacy
import pickle
```

```python
with open('gutenberg_data.pkl','rb') as f:
    gutenberg_df = pickle.load(f)
```

```python
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner"])
```

```python
def tag_full_document(text):
    '''
    Retrieve part of speech tags for a full document
    '''
    doc = nlp(text)
    return [[(token.text, token.pos_) for token in sent] for sent in doc.sents]
```

```python
gutenberg_df.head()
```

```python
tagged_sentences = []
for i in range(gutenberg_df.shape[0]):
    doc = gutenberg_df.iloc[i]['Text']
    try:
       tagged_sentences.append(tag_full_document(doc))
       print(f'Tagged document {i} of {gutenberg_df.shape[0]}')
    except Exception:
        print(f'Failed to tag document {i} of {gutenberg_df.shape[0]}')
```

```python
with open('tagged_documents.pkl','wb') as f:
    pickle.dump(tagged_sentences, f)
```

```python
flattened_tagged_sentences = [sent for doc in tagged_sentences for sent in doc]

sampled_idxes = np.random.choice(len(flattened_tagged_sentences), 10000, replace=False)
sampled_sentences = [flattened_tagged_sentences[i] for i in sampled_idxes]
```

```python
with open('tagged_sentences.pkl','wb') as f:
    pickle.dump(sampled_sentences, f)
```
